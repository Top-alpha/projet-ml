# Découpage et protocole d'évaluation creation de notre modele de machine learning

<h2>1__DECOUPAGE DES DONNEES</h2>

<p>
    Avant de procéder à l'entraînement des modèles de classification, il est nécessaire de diviser le jeu de données en plusieurs sous-ensembles distincts. Ce découpage permet d'évaluer de manière fiable les performances des modèles tout en limitant les risques de surapprentissage. Dans cette étude, le dataset a été séparé en trois ensembles : un ensemble d'entraînement destiné à l'apprentissage du modèle, un ensemble de validation utilisé pour l'ajustement et la comparaison des modèles, ainsi qu'un ensemble de test réservé à l'évaluation finale. Afin de conserver la distribution initiale des classes, une stratification ((Conservation du meme quantité de train, validation et test)) a été appliquée lors du découpage des données. La répartition retenue est de 70 % pour l'entraînement, 20 % pour la validation et 10 % pour le test.
</p>


```python

from sklearn.model_selection import train_test_split
import pandas as dat

df= dat.read_csv("../data/processed/dataset_clean.csv")
# Longueur calculée à partir de l'URL
# url_length_calculee = df["URL"].str.len()
# df["URLLength"] = df["URL"].str.len()
# # Lignes où URLLength est incorrect
# erreurs = df[df["URLLength"] != url_length_calculee]
# print(df["label"].value_counts())
# print(f"Nombre d'erreurs : {len(erreurs)}")
# print(erreurs[["URL", "URLLength"]].head())

# Séparation des features (X) et de la cible (y) en d'autres termes les features sont ce que le createur du modele donne a ce dernier pour qu'il parviennent a deduire les cible y qui est comme je l'ai dit ce que le modele doit savoir predire
# donc
colonnes_a_supprimer = [
    "label", 
    "URL",
    "Domain"
]
Caract = df.drop(columns=colonnes_a_supprimer)#, axis=1)
Cible = df["label"]

X_train, X_test, y_train, y_test = train_test_split(
    Caract, Cible, test_size=0.1, random_state=42, stratify=Cible
)
## Division du train en train et validation
X_train, X_val, y_train, y_val = train_test_split(
     X_train, y_train, test_size=0.2222, random_state=42, stratify=y_train
 )
# Ici : 70% train, 20% validation, 10% test

print("Train :", X_train.shape)
print("Validation :", X_val.shape)
print("Test :", X_test.shape)
```

<p>Le découpage effectué a conduit aux sous-ensembles suivants :
<ul>
<li>Ensemble d'entraînement :  164763 observations et 18 variables ;</li>
<li>Ensemble de validation : 47070 observations et 18 variables ;</li>
<li>Ensemble de test : 23537 observations et 18 variables.</li>
</ul>
Cette répartition garantit que la majorité des données est utilisée pour l'apprentissage tout en conservant suffisamment d'observations pour la validation et l'évaluation finale des modèles.</p>

<h2>2 __ PROTOCOLE D'EVALUATION</h2>

<p>
Avant de procéder à l'entraînement des modèles de classification, il est nécessaire de diviser le jeu de données en plusieurs sous-ensembles distincts. Ce découpage permet d'obtenir une évaluation fiable des performances des modèles tout en limitant les risques de surapprentissage. L'objectif est de mesurer la capacité du modèle à généraliser sur de nouvelles données qu'il n'a jamais observées auparavant.
</p>

<p>
Le jeu de données a été séparé en trois ensembles distincts :
</p>

<ul>
    <li><strong>Ensemble d'entraînement :</strong> 164763 observations (70 %)</li>
    <li><strong>Ensemble de validation :</strong> 47070 observations (20 %)</li>
    <li><strong>Ensemble de test :</strong> 23537 observations (10 %)</li>
</ul>

<p>
Une stratification a été appliquée lors du découpage afin de conserver la même distribution des classes dans chacun des sous-ensembles. Cette méthode garantit que les proportions d'URL légitimes et d'URL de phishing restent similaires dans les ensembles d'entraînement, de validation et de test.
</p>

<h4>2.1__ Rôle de chaque ensemble</h4>

<h5>Ensemble d'entraînement</h5>

<p>
L'ensemble d'entraînement est utilisé pour apprendre les relations entre les caractéristiques des URL et leur classe associée. Les algorithmes exploitent ces données afin d'ajuster leurs paramètres internes.
</p>

<h5>Ensemble de validation</h5>

<p>
L'ensemble de validation sert à comparer les différents modèles de Machine Learning ainsi qu'à ajuster leurs paramètres. Il permet de sélectionner le modèle présentant les meilleures performances avant l'évaluation finale.
</p>

<h5>Ensemble de test</h5>

<p>
L'ensemble de test est utilisé uniquement à la fin du processus. Les données qu'il contient ne sont jamais utilisées pendant l'entraînement ou la validation. Cet ensemble fournit une estimation réaliste des performances du modèle sur des données inédites.
</p>

<h4>2.2__ Protocole d'évaluation</h4>

<p>
L'évaluation des modèles suivra les étapes suivantes :
</p>

<ol>
    <li>Entraînement du modèle sur l'ensemble d'entraînement.</li>
    <li>Évaluation sur l'ensemble de validation.</li>
    <li>Comparaison des performances des différents modèles.</li>
    <li>Sélection du modèle le plus performant.</li>
    <li>Évaluation finale du modèle retenu sur l'ensemble de test.</li>
</ol>

<p>
Cette procédure garantit une comparaison équitable entre les modèles tout en préservant l'indépendance des données de test.
</p>

<h4>2.3__ Métriques d'évaluation</h4>

<p>
Pour mesurer les performances des modèles, plusieurs indicateurs seront utilisés :
</p>

<ul>
    <li><strong>Accuracy :</strong> proportion de prédictions correctes parmi l'ensemble des prédictions réalisées.</li>
    <li><strong>Precision :</strong> proportion d'URL prédites comme malveillantes qui le sont réellement.</li>
    <li><strong>Recall :</strong> capacité du modèle à détecter les URL de phishing présentes dans le jeu de données.</li>
    <li><strong>F1-Score :</strong> compromis entre la précision et le rappel permettant une évaluation globale des performances.</li>
</ul>

<h4>2.4__ Prévention du surapprentissage</h4>

<p>
Afin de limiter le risque de surapprentissage, les ensembles de validation et de test sont strictement séparés de l'ensemble d'entraînement. Les performances finales seront calculées uniquement sur l'ensemble de test afin d'obtenir une évaluation impartiale et représentative du comportement du modèle dans un contexte réel.
</p>

<h6>3__ Stockage des données d'entrainment sur mon splits</h6>

```python
import pandas as pd

# Reconstitution des datasets

train_df = pd.concat([X_train, y_train], axis=1)
val_df = pd.concat([X_val, y_val], axis=1)
test_df = pd.concat([X_test, y_test], axis=1)

# Sauvegarde

train_df.to_csv("../data/splits/train.csv", index=False)
val_df.to_csv("../data/splits/validation.csv", index=False)
test_df.to_csv("../data/splits/test.csv", index=False)

print("Fichiers enregistrés avec succès")
```

Ce protocole garantit une mesure fiable des performances des futurs modèles de détection d'URL de phishing.
